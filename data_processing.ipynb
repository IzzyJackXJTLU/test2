```python
import pandas as pd
import chardet

# 检测文件编码
with open('jd1.csv', 'rb') as f:
    result = chardet.detect(f.read())
    encoding = result['encoding']
    print(f"文件编码格式为: {result['encoding']}")

# 使用检测到的编码读取
df = pd.read_csv('jd1.csv', encoding=encoding)
df
```

```python
#删除数据集里面的重复行
df.drop_duplicates(inplace=True)
df.to_csv('jd1.csv',index=False, encoding='utf-8-sig')
df
```
